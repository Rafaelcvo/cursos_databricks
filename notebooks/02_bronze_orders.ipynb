```python
# =========================
# Paths
# =========================
raw_path = '/Volumes/main/default/novamarket/raw/olist_order_payments_dataset/'
```

```python
# =========================
# Leitura CSV
# =========================

df_orders = (
    spark
    .read
    .format('csv')
    .option('header', 'true')
    .option('inferSchema', 'true')
    .load(raw_path)
)

spark.sql('''
          CREATE SCHEMA IF NOT EXISTS bronze
          ''')

# =========================
# Salvar como Delta Table
# =========================
(
    df_orders.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("bronze.orders")
)

print("Tabela bronze.orders criada com sucesso.")
```
